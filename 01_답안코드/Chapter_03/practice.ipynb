실습: OutPutParser 활용하기

요구사항 1
1. 사용자의 질문에 대해 답변을 생성하는 체인을 작성하시오.
2. 체인은 아래와 같이 동작해야 합니다:
    * 입력 데이터: 사용자의 질문 (예: "인셉션 감독이 누구인가요?")
    * 출력 데이터: 질문에 대한 답변 answer (예 : "영화 "인셉션"의 감독은 크리스토퍼 놀란(Christopher Nolan)입니다.")
3. 답변을 생성하기 위해 다음 구성 요소를 사용하시오:
    * PromptTemplate: 사용자의 질문을 기반으로 적절한 답변을 생성하는 템플릿을 작성하시오.
    * ChatOpenAI: OpenAI 모델로 답변을 생성하시오.
    * StrOutputParser: 답변을 문자열로 반환하시오.
4. 이 체인을 chain1으로 정의하고, invoke 메서드를 사용하여 답변을 answer 변수에 저장하시오.


```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
```

```python
question = "인셉션 감독이 누구인가요?"
```

```python
template1 = """
당신은 영화를 추천해주는 AI 챗봇입니다. User의 질문에 대해 답변하시오.
question : {question}
"""

prompt1 = PromptTemplate(
    template=template1
)

model1 = ChatOpenAI(model_name="gpt-5-nano")

chain1 = prompt1 | model1 | StrOutputParser()

answer = chain1.invoke({"question" : question})
```

```python
print('출력 타입 : ', type(answer))
print('출력 결과 : ', answer)
```

요구사항 2
1. 요구사항 1에서 생성된 answer와 사용자의 질문을 기반으로, 사용자가 다음에 질문할 만한 추천 프롬프트 3개를 리스트로 생성하시오.
2. 체인은 아래와 같이 동작해야 합니다:
    * 입력 데이터:
        * question: 사용자의 최초 질문. (인셉션 감독이 누구인가요?)
        * answer: 요구사항 1에서 생성된 답변.
    * 출력 데이터: 사용자가 질문할 만한 추천 프롬프트 3개의 리스트 result (예: ['인셉션과 비슷한 영화는 뭐가 있나요?', '크리스토퍼 놀란의 다른 작품 추천해 주세요', '인셉션에 출연한 배우들이 나온 영화는?'])
3. 추천 프롬프트는 다음 구성 요소를 사용하여 생성하시오:
    * PromptTemplate: 질문과 답변을 기반으로 새로운 추천 프롬프트를 생성하는 템플릿을 작성하시오.
    * ChatOpenAI: OpenAI 모델로 추천 프롬프트를 생성하시오.
    * CommaSeparatedListOutputParser: 추천 프롬프트를 리스트 형식으로 반환하시오.
4. 이 체인을 chain2로 정의하고, invoke 메서드를 사용하여 결과를 result 변수에 저장하시오.

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser

# 콤마로 구분된 리스트 출력 파서 초기화
output_parser = CommaSeparatedListOutputParser()

# 출력 형식 지침
format_instructions = output_parser.get_format_instructions()
```

```python
template2 = """
당신은 영화를 추천해 주는 AI 챗봇입니다. User의 질문과 AI의 답변 다음으로 User가 질문할 만한 추천 프롬프트 3개를 말해주세요.
User_Message : {question}
AI_Message : {answer}

FORMAT :
{format}
"""

prompt2 = PromptTemplate(
    template=template2,
    partial_variables={
        "format": format_instructions
    },
)

model2 = ChatOpenAI(model_name="gpt-5-nano", temperature=1.2)

chain2 = prompt2 | model2 | output_parser

result = chain2.invoke({"question" : question, "answer" : answer})
```

```python
print('출력 타입 : ', type(answer))
print('출력 결과 : ', answer)
print('출력 타입 : ', type(result))
print('출력 결과 : ', result)
```
